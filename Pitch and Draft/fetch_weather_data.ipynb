```python
import requests
import pandas as pd

def fetch_wind(name, longitude, latitude, start_date, end_date):

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "start_date": start_date,
        "end_date": end_date,
        "hourly": "wind_speed_100m,wind_direction_100m",
        "timezone": "Pacific/Auckland"
    }

    response = requests.get(url, params=params)
    data = response.json()

    print(data)

    # Turn it into a DataFrame
    df = pd.DataFrame({
        "time": data["hourly"]["time"],
        "wind_speed_kmh": data["hourly"]["wind_speed_100m"],
        "wind_direction_deg": data["hourly"]["wind_direction_100m"]
    })

    df["time"] = pd.to_datetime(df["time"])
    print(df.head(10))
    print(f"\nShape: {df.shape}")
    print(f"Missing values: {df.isnull().sum().sum()}")

    df = df.rename(columns={
    'wind_speed_kmh': f'{name}_wind_kmh',
    'wind_direction_deg': f'{name}_wind_direction_deg'
    })
    return df
```

```python

#             name,             longitude,latitude, start_date, end_date
request =  [['Timaro'           , 171.49, -44.39, "2021-01-01", "2026-04-30"],
            ['Southern_Tip'     , 168.11, -46.11, "2021-01-01", "2026-04-30"],
            ['Wellington'       , 174.66, -41.25, "2021-01-01", "2026-04-30"],
            ['Palmerston North' , 175.76, -40.34, "2021-01-01", "2026-04-30"],
            ['Raglan'           , 174.95, -37.86, "2021-01-01", "2026-04-30"]]

wind_data = pd.DataFrame()
for name, lon, lat, start, end in request:
    df = fetch_wind(name, lon, lat, start, end)

    if wind_data.empty:
        wind_data = df
    else:
        wind_data = wind_data.merge(df, on='time', how='outer')

print(wind_data.head(5))
```
