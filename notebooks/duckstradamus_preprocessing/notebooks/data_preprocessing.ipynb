```python
# load data
file_path = "../dataset/Lakes storage levels/NI_TPO_Storage_LakeTaupo.csv"

lake_storage = pd.read_csv(file_path)
lake_storage.head()
```

```python
# basic inspection
lake_storage.info()
lake_storage.isnull().sum()
lake_storage.duplicated().sum()
```

```python
# clean the data columns
lake_storage.columns = (
    lake_storage.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)

lake_storage.columns
```

```python
# create datetime column
lake_storage["datetime"] = pd.to_datetime(
    lake_storage["date"] + " " + lake_storage["time"],
    errors="coerce"
)
```

```python
# keep only useful columns
lake_storage = lake_storage[
    ["datetime", "lake_level_m", "active_storage_mm³"]
]
```

```python
# drop missing values
lake_storage.dropna(inplace=True)
```

```python
# remove duplicates
lake_storage.drop_duplicates(inplace=True)
```

```python
# sort by the datetime
lake_storage = lake_storage.sort_values("datetime")
```

```python
# reset index
lake_storage.reset_index(drop=True, inplace=True)
```

```python
# final check
lake_storage.info()
lake_storage.head()
lake_storage.isnull().sum()
```

```python
# =========================
# SAVE CLEANED FILES
# =========================

from pathlib import Path

# create cleaned_data folder if it doesn't exist
output_folder = Path("../cleaned_data")
output_folder.mkdir(exist_ok=True)

# dictionary of all lakes
lake_files = {
    "lake_taupo": "../dataset/Lakes storage levels/NI_TPO_Storage_LakeTaupo.csv",
    "lake_waikaremoana": "../dataset/Lakes storage levels/NI_WKA_Storage_LakeWaikaremoana.csv",
    "lake_hawea": "../dataset/Lakes storage levels/SI_HWE_Storage_LakeHawea.csv",
    "lake_manapouri": "../dataset/Lakes storage levels/SI_MAN_Storage_LakeManapouri.csv",
    "lake_ohau": "../dataset/Lakes storage levels/SI_OHA_Storage_LakeOhau.csv",
    "lake_pukaki": "../dataset/Lakes storage levels/SI_PKI_Storage_LakePukaki.csv",
    "lake_teanau": "../dataset/Lakes storage levels/SI_TAU_Storage_LakeTeAnau.csv",
    "lake_tekapo": "../dataset/Lakes storage levels/SI_TKA_Storage_LakeTekapo.csv",
    "lake_wanaka": "../dataset/Lakes storage levels/SI_WAN_Storage_LakeWanaka.csv",
    "lake_wakatipu": "../dataset/Lakes storage levels/SI_WPU_Storage_LakeWakatipu.csv",
}

# store cleaned dfs
cleaned_lakes = {}

# loop through all files
for lake_name, file_path in lake_files.items():

    # preprocess
    df = preprocess_lake_storage(file_path)

    # store dataframe
    cleaned_lakes[lake_name] = df

    # save cleaned csv
    save_path = output_folder / f"{lake_name}_cleaned.csv"

    df.to_csv(save_path, index=False)

    print(f"✅ Saved successfully: {save_path}")

print("\n🎉 All lake files cleaned and saved!")
```
